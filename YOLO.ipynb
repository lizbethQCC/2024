```python
!pip install ultralytics
from ultralytics import YOLO

# Cargar modelo (se descarga solo la primera vez)
modelo = YOLO('yolov8n.pt')

# Pon aquí la ruta de tu imagen: 'casa.jpg' o 'foto.png'
resultados = modelo('https://ultralytics.com/images/bus.jpg')

# Guarda la imagen con los recuadros dibujados
resultados[0].save('resultado.jpg')

print("¡Listo! Revisa el archivo 'resultado.jpg'")
```
